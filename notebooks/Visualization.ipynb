```python
# NOTE ON COLUMNS:
# The Kaggle files sometimes use '\N' for missing numbers. We convert
# the points/position columns to real numbers so we can add them up.
results["points"]   = pd.to_numeric(results["points"], errors="coerce")
results["grid"]     = pd.to_numeric(results["grid"], errors="coerce")
```

### **Data Integration - join everything into ONE big table**

```python
f1 = results.merge(drivers,      on="driverId",      how="left")
f1 = f1.merge(constructors,      on="constructorId", how="left",
              suffixes=('_driver', '_constructor'))   # both tables have 'name'/'nationality'
f1 = f1.merge(races,             on="raceId",        how="left")
f1 = f1.merge(circuits,          on="circuitId",     how="left",
              suffixes=('_race', '_circuit'))          # both have 'name'

# After merging, pandas renames clashing columns. To keep things clear,
# we build a few tidy columns with obvious names:
f1["driver_name"]      = f1["forename"].astype(str) + " " + f1["surname"].astype(str)
f1["constructor_name"] = f1["name_x"]      # team name, corrected to use name_x from constructors
f1["circuit_country"]  = f1["country"]               # from circuits.csv
f1["season"]           = f1["year"]                  # from races.csv
```

### **Examine the dataset**

```python
pd.set_option('display.max_columns', None)

# Rename columns for clarity
f1 = f1.rename(columns={
    'number_x': 'result_driver_number',
    'number_y': 'driver_number',
    'time_x': 'driver_race_time',
    'time_y': 'race_start_time',
    'name_y': 'race_name',
    'name': 'circuit_name' # This 'name' is from circuits.csv
})

# Drop unnecessary and redundant columns
columns_to_drop = [
    'forename', 'surname', # Redundant as 'driver_name' exists
    'name_x', # Redundant as 'constructor_name' exists
    'year',   # Redundant as 'season' exists
    'country', # Redundant as 'circuit_country' exists

    # URL columns as requested
    'url_driver',
    'url_constructor',
    'url_race',
    'url_circuit',

    # Foreign key for races table is raceId, and other attributes are already taken out.
    # fp1_date, fp1_time, fp2_date, fp2_time, fp3_date, fp3_time, quali_date, quali_time, sprint_date, sprint_time
    # are all specific session times, which might be useful, so I will keep them for now.
]

# Ensure we only try to drop columns that actually exist
columns_to_drop = [col for col in columns_to_drop if col in f1.columns]
f1 = f1.drop(columns=columns_to_drop)

# Reposition columns - define a desired order for key columns
# Start with IDs, then key info, then race details, then driver/constructor details, then result details.

# Get current columns
current_columns = f1.columns.tolist()

# Define a desired order for some important columns
desired_order_prefix = [
    'resultId', 'raceId', 'driverId', 'constructorId', 'circuitId',
    'season', 'round', 'race_name', 'date', 'race_start_time',
    'circuit_name', 'location', 'circuit_country', 'lat', 'lng', 'alt', 'circuitRef',
    'driver_name', 'driver_number', 'code', 'nationality_driver', 'dob', 'driverRef',
    'constructor_name', 'nationality_constructor', 'constructorRef',
    'grid', 'position', 'positionText', 'positionOrder',
    'points', 'laps', 'driver_race_time', 'milliseconds',
    'fastestLap', 'rank', 'fastestLapTime', 'fastestLapSpeed', 'statusId',
    'result_driver_number' # Renamed from number_x
]

# Filter desired_order_prefix to only include columns that still exist in f1
desired_order_prefix = [col for col in desired_order_prefix if col in f1.columns]

# Get the remaining columns not in desired_order_prefix
remaining_columns = [col for col in current_columns if col not in desired_order_prefix]

# Combine to create the final column order, placing remaining columns at the end
final_column_order = desired_order_prefix + remaining_columns

# Apply the new column order
f1 = f1[final_column_order]

print("First 5 rows (after cleanup and reordering):")
display(f1.head())

print("\nDataset info (after cleanup and reordering):")
display(f1.info())

print("\nStatistical summary of points (after cleanup and reordering):")
display(f1[["points", "grid"]].describe())

print("\nMissing values per key column (after cleanup and reordering):")
display(f1[["points", "driver_name", "constructor_name", "season"]].isnull().sum())

print("\nUnique constructors (after cleanup and reordering):", f1["constructor_name"].nunique())
print("Unique drivers (after cleanup and reordering):", f1["driver_name"].nunique())
print("Seasons covered (after cleanup and reordering):", f1["season"].min(), "to", f1["season"].max())
```

### **Bar Chart**

```python
# ---------------------------------------------------------------
# Top 10 Drivers by Total Points
# ---------------------------------------------------------------
driver_points = (
    f1.groupby("driver_name")["points"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 5))
driver_points.plot(kind="barh", color="steelblue")
plt.title("Top 10 Drivers by Total Career Points")
plt.xlabel("Driver")
plt.ylabel("Total Points")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------------
# Points by Driver Nationality
# ---------------------------------------------------------------
nationality_points = (
    f1.groupby("nationality_driver")["points"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 5))
nationality_points.plot(kind="bar", color="darkorange")
plt.title("Total Points by Driver Nationality (Top 10)")
plt.xlabel("Nationality")
plt.ylabel("Total Points")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------------
# Number of Race Entries per Constructor (Top 10)
# ---------------------------------------------------------------
constructor_entries = (
    f1.groupby("constructor_name")["resultId"]
      .count()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 5))
constructor_entries.plot(kind="bar", color="seagreen")
plt.title("Number of Race Entries per Constructor (Top 10)")
plt.xlabel("Constructor")
plt.ylabel("Race Entries")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# ---------------------------------------------------------------
# Points by Circuit Country (Top 10)
# ---------------------------------------------------------------
country_points = (
    f1.groupby("circuit_country")["points"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 5))
country_points.plot(kind="bar", color="mediumpurple")
plt.title("Total Points Scored by Circuit Country (Top 10)")
plt.xlabel("Country")
plt.ylabel("Total Points")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### **Pie Chart**

```python
# ---------------------------------------------------------------
# Points Share of Top 6 Constructors
# ---------------------------------------------------------------
constructor_points = (
    f1.groupby("constructor_name")["points"]
      .sum()
      .sort_values(ascending=False)
      .head(6)
)

plt.figure(figsize=(5, 5))
plt.pie(
    constructor_points.values,
    labels=constructor_points.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Points Share of Top 6 Constructors")
plt.tight_layout()
plt.show()
```

### **Line Chart**

```python
# ---------------------------------------------------------------
# Total Points Awarded per Season (Trend Over Time)
# ---------------------------------------------------------------
season_points = (
    f1.groupby("season")["points"]
      .sum()
      .sort_index()
)

plt.figure(figsize=(11, 5))
plt.plot(season_points.index, season_points.values, marker="o")
plt.title("Total Points Awarded per Season")
plt.xlabel("Season (Year)")
plt.ylabel("Total Points")
plt.grid(True)
plt.tight_layout()
plt.show()
```

### **Grouped Bar Chart**

```python
# ---------------------------------------------------------------
# Points by Constructor across a few Seasons
# ---------------------------------------------------------------
recent = f1[f1["season"].isin([2021, 2022, 2023, 2024])]
top_teams = (
    recent.groupby("constructor_name")["points"].sum()
          .sort_values(ascending=False).head(5).index
)
recent = recent[recent["constructor_name"].isin(top_teams)]

pivot = pd.pivot_table(
    recent,
    values="points",
    index="season",
    columns="constructor_name",
    aggfunc="sum",
    fill_value=0
)

pivot.plot(kind="bar", figsize=(11, 5))
plt.title("Constructor Points by Season (2021-2024, Top 5 Teams)")
plt.xlabel("Season")
plt.ylabel("Total Points")
plt.xticks(rotation=0)
plt.legend(title="Constructor")
plt.tight_layout()
plt.show()
```

### **Heatmap**

```python
# ---------------------------------------------------------------
# Correlation Between Numerical Attributes
# ---------------------------------------------------------------
numerical = f1[["grid", "points", "season"]].dropna()
corr = numerical.corr()

plt.figure(figsize=(6, 5))
img = plt.imshow(corr, interpolation="nearest", cmap="coolwarm")
plt.colorbar(img)
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.index)), corr.index)
for i in range(len(corr.index)):
    for j in range(len(corr.columns)):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")
plt.title("Correlation Heatmap (grid, points, season)")
plt.tight_layout()
plt.show()
```
